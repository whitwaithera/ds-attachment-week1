```python
import pandas as pd
import numpy as np

# Load the raw dataset (it's in the same folder)
df = pd.read_csv('diabetes_risk.csv')

print("Shape:", df.shape)
print("\nData types:\n", df.dtypes)
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Fill missing values with "Unknown" to preserve data
df['smoking_status']      = df['smoking_status'].fillna('Unknown')
df['alcohol_consumption'] = df['alcohol_consumption'].fillna('Unknown')
df['income_bracket']      = df['income_bracket'].fillna('Unknown')

print("Remaining missing cells:", df.isnull().sum().sum())
```

```python
cat_cols = ['gender','city','family_history_diabetes',
            'physical_activity_level','diet_type','smoking_status',
            'alcohol_consumption','income_bracket','diabetes_risk']

for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.title()

print("Categorical columns standardized.")
```

```python
# Coerce numbers
num_cols = ['age','bmi','hours_sleep_per_night','stress_level',
            'fasting_blood_sugar','hba1c_level','blood_pressure_systolic',
            'blood_pressure_diastolic','waist_circumference_cm']
for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# Define clinical ranges and filter outliers
ranges = {
    'age': (0, 120),
    'bmi': (10, 70),
    'hours_sleep_per_night': (0, 24),
    'stress_level': (1, 10),
    'fasting_blood_sugar': (50, 500),
    'hba1c_level': (3, 15),
    'blood_pressure_systolic': (70, 250),
    'blood_pressure_diastolic': (40, 150),
    'waist_circumference_cm': (40, 200),
}

start_rows = len(df)
for col, (lo, hi) in ranges.items():
    df = df[(df[col] >= lo) & (df[col] <= hi)]

print(f"Dropped {start_rows - len(df)} outlier rows.")
```

```python
# Pulse pressure
df['pulse_pressure'] = df['blood_pressure_systolic'] - df['blood_pressure_diastolic']

# BMI category
def bmi_cat(b):
    if b < 18.5:  return 'Underweight'
    if b < 25:    return 'Normal'
    if b < 30:    return 'Overweight'
    return 'Obese'
df['bmi_category'] = df['bmi'].apply(bmi_cat)

# Age group
df['age_group'] = pd.cut(df['age'], bins=[0, 30, 45, 60, 120], labels=['<30', '30-45', '45-60', '60+'])
```

```python
# Save the cleaned dataset into the same folder
df.to_csv('diabetes_risk_clean.csv', index=False)
print("✅ Successfully saved diabetes_risk_clean.csv")
print("Final shape:", df.shape)
```
